```python
class Planet:
    def __init__(
        self, name, planet_type, star
    ):  # __init__ spesial method yanng dipanggil pertama kali ketika instansiasi class
        self.name = name
        self.planet_type = planet_type
        self.star = star
        if not isinstance(
            name, str
        ):  # isinstance adalah fungsi Python untuk memeriksa apakah suatu nilai/object merupakan instance dari tipe atau class tertentu.
            raise TypeError("name, planet type, and star must be strings")
        if not isinstance(planet_type, str):
            raise TypeError("name, planet type, and star must be strings")
        if not isinstance(star, str):
            raise TypeError("name, planet type, and star must be strings")
        if name == "":
            raise ValueError("name, planet_type, and star must be non-empty strings")
        if planet_type == "":
            raise ValueError("name, planet_type, and star must be non-empty strings")
        if star == "":
            raise ValueError("name, planet_type, and star must be non-empty strings")

    def orbit(self):
        return f"{self.name} is orbiting around {self.star}..."

    def __str__(
        self,
    ):  # tidak perlu dipanggil secara eksplisit karena otomatis dipanggil pertama selama instansiasi class.
        return f"Planet: {self.name} | Type: {self.planet_type} | Star: {self.star}"


planet_1 = Planet("Neptunus", "bulat", "Andromeda")
planet_2 = Planet("Uranus", "kotak", "Alpha Centauri")
planet_3 = Planet("Pluto", "lonjong", "Epsilon Eridani")

print(planet_1)  # sama dengan planet_1.__str__
print(planet_2)
print(planet_3)

print(planet_1.orbit())
print(planet_2.orbit())
print(planet_3.orbit())
```
