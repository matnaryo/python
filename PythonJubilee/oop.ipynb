```python
# Class, method dan attribute
class Mobil:
    def __init__(self, warna, merk, kecepatan):
        self.warna = warna
        self.merk = merk
        self.kecepatan = kecepatan

    def tambah_kecepatan(self):
        self.kecepatan += 10


mobil_1 = Mobil("Merah", "Mazda", 100)
print("Sebelum ditambahkan: ")
print(mobil_1.kecepatan)

mobil_1.tambah_kecepatan()

print(f"Setelah ditambahkan:,  {mobil_1.kecepatan}")
print(mobil_1.kecepatan)
```
