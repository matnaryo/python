# **Method to Access List**


```python
list_developer = ["Rohmat", 25, ["Golang", "Python", "PHP"]]
list_developer[2][1]
name, age, coding = list_developer
print(name)
print(age)
print(coding)
```

```python
# metode akses list

# append
numbers = [1, 2, 3, 4, 5, 6]
numbers.append(7)
print(numbers)
```

```python
even_numbers = [8, 9, 10]
numbers.append(even_numbers)
numbers.extend(even_numbers)
print(numbers)
```

Perhatikan dengan append tambahan dari proses `append` akan dikelompokkan dalam satu indeks. Berbeda dengan `extend`, maka setiap elemen dalam variabel tambahan akan dibongkar dan menjadi indeks yang berurutan.


```python
# insert
from numpy import number

numbers.insert(7, 7.5)
print(numbers)
```

```python
numbers.remove([8, 9, 10])
numbers.remove([8, 9, 10])
print(numbers)
```

```python
# pop untuk remove berdasarkan index. Jika disebutkan secara spesifik, maka indeks terakhir yang akan dihapus
numbers.pop(7)
print(numbers)
```

```python
# clear untuk mengosonghkan list

numbers.clear()
print(numbers)
```

```python
numbers = [3, 2, 5, 6, 4, 1]
numbers.sort()
print(numbers)
```

# **Tuple**

`Tuple` adalah tipe data yang menyimpan suatu nilai berurutan. `Tuple` dapat berisi campuran tipe data. yang membedakan dengan `list` adalah bahwa `tuple` bersifat immutable atau tidak dapat dirubah.


```python
developer = ("Rohmat Sunaryo", 42, "Pawang Python")
print(developer[2])
```

```python
# cara lain membuat tuple dari string
developer = "Rohmat Sunaryo"
tuple(developer)
"R" in developer
```
